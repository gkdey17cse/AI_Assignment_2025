## This notebook will create different training & Testing data for LLMs & Test their Resoponse

### 1. Iris Dataset

```python
# Import libraries
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.utils import shuffle

# Step 1: Load the Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# Convert to DataFrame
df = pd.DataFrame(X, columns=iris.feature_names)
df['target'] = y

# Step 2: Shuffle the data very well
df = shuffle(df, random_state=42).reset_index(drop=True)
# Print shape after shuffle
print("Shuffled dataset shape:", df.shape)
```

```python

# Step 3: Split into training (100) and testing (50)
train_df = df.iloc[:100]
test_df = df.iloc[100:150]

# Create the test file without target
test_no_target_df = test_df.drop(columns=['target'])

# Step 4: Save to CSV files
train_df.to_csv("../Data/iris_train.csv", index=False)
test_df.to_csv("../Data/iris_test.csv", index=False)
test_no_target_df.to_csv("../Data/iris_test_no_target.csv", index=False)

print("Saved iris_train.csv (100 rows), iris_test.csv (50 rows), iris_test_no_target.csv (50 rows).")
```

# Response Checker

### Adult Dataset

```python

import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score


def load_and_prepare_data(actual_file, predicted_file):
    """Load and preprocess actual and predicted datasets"""

    df_actual = pd.read_csv(actual_file)
    df_pred = pd.read_csv(predicted_file)

    print("Loaded Actual Shape:", df_actual.shape)
    print("Loaded Predicted Shape:", df_pred.shape)

    # Clean any trailing spaces, dots, weird formatting in income column
    def clean_income(val):
        val = str(val).strip().replace(".", "")
        if val in [">50K", ">50K"]:
            return 1
        return 0

    df_actual["income_binary"] = df_actual["income"].apply(clean_income)
    df_pred["income_binary"] = df_pred["income"].apply(clean_income)

    return df_actual["income_binary"].values, df_pred["income_binary"].values



def evaluate(y_true, y_pred):

    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)

    return accuracy, precision, recall, f1



def main():
    print("\nEvaluating Model...\n")

    # -------------------------------------------------------
    # OPTION 1: RELATIVE PATHS (Recommended)
    # -------------------------------------------------------
    actual_file = "../Data/adult_test.csv"
    predicted_file = "../Data/adult_llm_response.csv"

    # -------------------------------------------------------
    # OPTION 2: ABSOLUTE PATHS (If you want full paths)
    # UNCOMMENT IF YOU WANT TO USE THESE:
    # actual_file = r"E:\MTechCSE\Study\Sem3\AI\Assignment\Assignment_4\Data\adult_test.csv"
    # predicted_file = r"E:\MTechCSE\Study\Sem3\AI\Assignment\Assignment_4\Data\adult_llm_response.csv"
    # -------------------------------------------------------

    y_true, y_pred = load_and_prepare_data(actual_file, predicted_file)

    # Ensure equal length
    min_len = min(len(y_true), len(y_pred))
    y_true = y_true[:min_len]
    y_pred = y_pred[:min_len]

    accuracy, precision, recall, f1 = evaluate(y_true, y_pred)

    print("\n==================== RESULTS ====================")
    print(f"Test Accuracy : {accuracy:.4f}")
    print(f"Precision     : {precision:.4f}")
    print(f"Recall        : {recall:.4f}")
    print(f"F1-Score      : {f1:.4f}")
    print("=================================================\n")



if __name__ == "__main__":
    main()
```

## Iris Dataset

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score


def load_and_prepare_data(actual_file, predicted_file):
    """Load actual and predicted Iris datasets and extract target column"""

    df_actual = pd.read_csv(actual_file)
    df_pred = pd.read_csv(predicted_file)

    print("Loaded Actual Shape:", df_actual.shape)
    print("Loaded Predicted Shape:", df_pred.shape)

    # Extract target labels
    y_true = df_actual["target"].astype(int).values
    y_pred = df_pred["target"].astype(int).values

    return y_true, y_pred



def evaluate(y_true, y_pred):

    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, average="macro", zero_division=0)
    recall = recall_score(y_true, y_pred, average="macro", zero_division=0)
    f1 = f1_score(y_true, y_pred, average="macro", zero_division=0)

    return accuracy, precision, recall, f1



def main():

    print("\nEvaluating Iris Predictions...\n")

    # -------------------------------------------------------
    # USE RELATIVE PATHS (recommended)
    # -------------------------------------------------------
    actual_file = "../Data/iris_test.csv"
    predicted_file = "../Data/iris_llm_response.csv"

    # -------------------------------------------------------
    # If you want absolute paths, uncomment:
    # actual_file = r"E:\MTechCSE\Study\Sem3\AI\Assignment\Assignment_4\Data\iris_actual.csv"
    # predicted_file = r"E:\MTechCSE\Study\Sem3\AI\Assignment\Assignment_4\Data\iris_predicted.csv"
    # -------------------------------------------------------

    y_true, y_pred = load_and_prepare_data(actual_file, predicted_file)

    # Match length
    min_len = min(len(y_true), len(y_pred))
    y_true = y_true[:min_len]
    y_pred = y_pred[:min_len]

    accuracy, precision, recall, f1 = evaluate(y_true, y_pred)

    print("\n==================== RESULTS ====================")
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1-Score : {f1:.4f}")
    print("=================================================\n")



if __name__ == "__main__":
    main()
```
