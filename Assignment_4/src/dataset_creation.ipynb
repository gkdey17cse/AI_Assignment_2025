## This notebook will create different training & Testing data for LLMs 

```python
# Import libraries
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.utils import shuffle

# Step 1: Load the Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# Convert to DataFrame
df = pd.DataFrame(X, columns=iris.feature_names)
df['target'] = y

# Step 2: Shuffle the data very well
df = shuffle(df, random_state=42).reset_index(drop=True)

# Step 3: Split into training (100) and testing (50)
train_df = df.iloc[:100]
test_df = df.iloc[100:150]

# Create the test file without target
test_no_target_df = test_df.drop(columns=['target'])

# Step 4: Save to CSV files
train_df.to_csv("../Data/iris_train.csv", index=False)
test_df.to_csv("../Data/iris_test.csv", index=False)
test_no_target_df.to_csv("../Data/iris_test_no_target.csv", index=False)

print("Saved iris_train.csv (100 rows), iris_test.csv (50 rows), iris_test_no_target.csv (50 rows).")
```

### For Wine Dataset

```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.utils import shuffle

# Load the wine dataset
wine = load_wine()
X = wine.data
y = wine.target

# Convert to DataFrame
df = pd.DataFrame(X, columns=wine.feature_names)
df["target"] = y

# Print the shape before shuffling
print("Original dataset shape:", df.shape)

# Shuffle very well
df = shuffle(df, random_state=42).reset_index(drop=True)

# Print shape after shuffle
print("Shuffled dataset shape:", df.shape)

# Show top 5 rows
df.head()
```

```python

# Split sizes
train_df = df.iloc[:120]
test_df = df.iloc[120:178]

# Create test data without target
test_no_target_df = test_df.drop(columns=["target"])

# Save all three files
train_df.to_csv("../Data/wine_train.csv", index=False)
test_df.to_csv("../Data/wine_test.csv", index=False)
test_no_target_df.to_csv("../Data/wine_test_no_target.csv", index=False)

print("Saved:")
print(" - wine_train.csv (120 rows, with target)")
print(" - wine_test.csv (58 rows, with target)")
print(" - wine_test_no_target.csv (58 rows, WITHOUT target)")
```
